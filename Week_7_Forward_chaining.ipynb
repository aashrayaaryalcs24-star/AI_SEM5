```python
KB = [
    ("American", "Robert"),
    ("Enemy", "A", "America"),
    ("Owns", "A", "T1"),
    ("Missile", "T1")
]

GOAL = ("Criminal", "Robert")

print("=== Initial Knowledge Base ===")
for fact in KB:
    print(f"  - {fact}")

step = 1
while True:
    print(f"\n=== Iteration {step} ===")
    NEW_FACTS = []

    # Rule 1: Missile(x) -> Weapon(x)
    print("\nEvaluating Rule 1: Missile(x) -> Weapon(x)")
    for fact in KB:
        if fact[0] == "Missile":
            x = fact[1]
            new_fact = ("Weapon", x)
            if new_fact not in KB and new_fact not in NEW_FACTS:
                print(f"  Match found: Missile({x}) -> Derived: Weapon({x})")
                NEW_FACTS.append(new_fact)

    # Rule 2: Missile(x) AND Owns(A,x) -> Sells(Robert, x, A)
    print("Evaluating Rule 2: Missile(x) AND Owns(A, x) -> Sells(Robert, x, A)")
    for fact in KB:
        if fact[0] == "Missile":
            x = fact[1]
            if ("Owns", "A", x) in KB:
                new_fact = ("Sells", "Robert", x, "A")
                if new_fact not in KB and new_fact not in NEW_FACTS:
                    print(f"  Match found: Missile({x}) AND Owns(A, {x}) -> Derived: Sells(Robert, {x}, A)")
                    NEW_FACTS.append(new_fact)

    # Rule 3: Enemy(x, America) -> Hostile(x)
    print("Evaluating Rule 3: Enemy(x, America) -> Hostile(x)")
    for fact in KB:
        if fact[0] == "Enemy" and fact[2] == "America":
            x = fact[1]
            new_fact = ("Hostile", x)
            if new_fact not in KB and new_fact not in NEW_FACTS:
                print(f"  Match found: Enemy({x}, America) -> Derived: Hostile({x})")
                NEW_FACTS.append(new_fact)

    # Rule 4: American(p) AND Weapon(q) AND Sells(p, q, r) AND Hostile(r) -> Criminal(p)
    print("Evaluating Rule 4: American(p) AND Weapon(q) AND Sells(p, q, r) AND Hostile(r) -> Criminal(p)")
    for fact in KB:
        if fact[0] == "American":
            p = fact[1]
            for weapon in KB:
                if weapon[0] == "Weapon":
                    q = weapon[1]
                    for sale in KB:
                        if sale[0] == "Sells" and sale[1] == p and sale[2] == q:
                            r = sale[3]
                            if ("Hostile", r) in KB:
                                new_fact = ("Criminal", p)
                                if new_fact not in KB and new_fact not in NEW_FACTS:
                                    print(f"  Match found: American({p}) AND Weapon({q}) AND Sells({p}, {q}, {r}) AND Hostile({r}) -> Derived: Criminal({p})")
                                    NEW_FACTS.append(new_fact)

    if NEW_FACTS:
        print("\nNew Facts Added in this iteration:")
        for fact in NEW_FACTS:
            print(f"  + {fact}")
        KB.extend(NEW_FACTS)
    else:
        print("\nNo new facts could be generated in this iteration.")

    if GOAL in KB:
        print("\n===========================")
        print("GOAL PROVED SUCCESSFULLY!")
        print(f"{GOAL[1]} is indeed a {GOAL[0]}.")
        print("===========================")
        break

    if not NEW_FACTS:
        print("\n===========================")
        print("GOAL CANNOT BE PROVED!")
        print("===========================")
        break

    step += 1
```
